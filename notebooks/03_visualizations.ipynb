```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv("../data/processed/credit_risk_cleaned.csv")
```

```python
total_customers = len(df)

total_loan = df["loan_amnt"].sum()

average_income = df["person_income"].mean()

default_rate = df["loan_status"].mean() * 100

average_interest = df["loan_int_rate"].mean()

print(total_customers)
print(total_loan)
print(average_income)
print(default_rate)
print(average_interest)
```

```python
fig = px.bar(
    df["loan_intent"].value_counts().reset_index(),
    x="loan_intent",
    y="count",
    color="loan_intent",
    title="Loan Purpose Distribution"
)

fig.show()
```

```python
fig = px.histogram(
    df,
    x="loan_grade",
    color="loan_grade",
    title="Loan Grade Distribution"
)

fig.show()
```

```python
fig = px.histogram(
    df,
    x="person_age",
    nbins=30,
    title="Customer Age Distribution"
)

fig.show()
```

```python
fig = px.histogram(
    df,
    x="person_income",
    title="Income Distribution"
)

fig.show()
```

```python
fig = px.pie(
    df,
    names="person_home_ownership",
    title="Home Ownership"
)

fig.show()
```

```python
risk = (
    df.groupby("loan_grade")["loan_status"]
      .mean()
      .reset_index()
)

risk["loan_status"] *= 100

fig = px.bar(
    risk,
    x="loan_grade",
    y="loan_status",
    color="loan_grade",
    title="Default Rate by Loan Grade (%)"
)

fig.show()
```

```python
fig = px.scatter(
    df,
    x="loan_int_rate",
    y="loan_amnt",
    color="risk_label",
    title="Interest Rate vs Loan Amount"
)

fig.show()
```

```python
corr = df.select_dtypes(include="number").corr()

fig = px.imshow(
    corr,
    text_auto=True,
    title="Correlation Matrix"
)

fig.show()
```

```python
fig = px.bar(
    df["employment_category"].value_counts().reset_index(),
    x="employment_category",
    y="count",
    color="employment_category",
    title="Employment Category"
)

fig.show()
```

```python
fig = px.pie(
    df,
    names="loan_size",
    title="Loan Size Distribution"
)

fig.show()
```
